```python
import json
import pandas as pd
from scipy.stats import pearsonr, spearmanr
import matplotlib.pyplot as plt
```

```python
with open(r"D:\finding jobs notes and usc things\USC projects\FORTIS LAB\AIEDUReportPlace\0908\result\llama_scores.json", "r", encoding="utf-8") as f:
    llama_data = json.load(f)
with open(r"D:\finding jobs notes and usc things\USC projects\FORTIS LAB\AIEDUReportPlace\0908\result\mistral_score.json", "r", encoding="utf-8") as f:
    mistral_data = json.load(f)
with open(r"D:\finding jobs notes and usc things\USC projects\FORTIS LAB\AIEDUReportPlace\0908\result\MRBench_V1_score_result.json", "r", encoding="utf-8") as f:
    mrbench_data = json.load(f)
```

```python
df_llama = pd.DataFrame([
    {"conversation_history": d.get("conversation_history") or d.get("conversation_histroy"),
     "llama_score_df": d["points"] if "points" in d else d.get("llama_score")}
    for d in llama_data
])

df_mistral = pd.DataFrame([
    {"conversation_history": d.get("conversation_history") or d.get("conversation_histroy"),
     "mistral_score_df": d["points"] if "points" in d else d.get("mistral_score")}
    for d in mistral_data
])

df_mrbench = pd.DataFrame([
    {"conversation_history": d["conversation_history"],
     "llama_score_mr": d["llama_score"],
     "mistral_score_mr": d["mistral_score"]}
    for d in mrbench_data
])
```

```python
df_mrbenchdf = df_mrbench.merge(df_llama, on="conversation_history").merge(df_mistral, on="conversation_history")
df_mrbenchdf
```

```python
from scipy.stats import pearsonr, spearmanr
import pandas as pd

results = []

# 逐個維度（例如一共有 8 個維度）計算
n_dim = 8
df = df_mrbenchdf
for i in range(n_dim):
    # 取出這個維度的分數
    llama_mr = [x[i] for x in df["llama_score_mr"]]
    llama_df = [x[i] for x in df["llama_score_df"]]
    mistral_mr = [x[i] for x in df["mistral_score_mr"]]
    mistral_df = [x[i] for x in df["mistral_score_df"]]

    results.append({
        "dimension": i,
        "pearson_llama": pearsonr(llama_mr, llama_df)[0],
        "spearman_llama": spearmanr(llama_mr, llama_df)[0],
        "pearson_mistral": pearsonr(mistral_mr, mistral_df)[0],
        "spearman_mistral": spearmanr(mistral_mr, mistral_df)[0],
    })

results_df = pd.DataFrame(results)
print(results_df)

```
